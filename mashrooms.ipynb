```python
import pandas as pd
df = pd.read_csv("mushrooms.csv")
```

```python

x = df.drop("type", axis=1)
y = df["type"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
for col in x_train.columns:
    le = LabelEncoder()
    x_train[col] = le.fit_transform(x_train[col])
    x_test[col] = le.transform(x_test[col])

le_y = LabelEncoder()
y_train = le_y.fit_transform(y_train)
y_test = le_y.transform(y_test)
```

```python
classifier = DecisionTreeClassifier()
classifier.fit(x_train, y_train)
```

```python
prediction = classifier.predict(x_test)
```

```python
print("Accuracy :", accuracy_score(y_test, prediction))
print("\nClassification Report :")
print(classification_report(y_test, prediction))
print("\nConfusion Matrix :")
print(confusion_matrix(y_test, prediction))
```

```python
# ================================
# 1. IMPORTS
# ================================
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import LabelEncoder

# ================================
# 2. DATA LOAD
# ================================
df = pd.read_csv("mushrooms.csv")
x = df.drop("type", axis=1)      # ⚠️ "type" likho, "class" nahi
y = df["type"]

# ================================
# 3. ENCODING
# ================================
x_encoded = x.copy()
for col in x_encoded.columns:
    le = LabelEncoder()
    x_encoded[col] = le.fit_transform(x_encoded[col])

le_y = LabelEncoder()
y_encoded = le_y.fit_transform(y)

print("✅ Encoding done! Shape:", x_encoded.shape)

# ================================
# 4. GRID SEARCH
# ================================
param_grid = {
    'max_depth': [1, 2, 3, 5, 7, 10, 15, 20],
    'criterion': ['gini', 'entropy'],
    'min_samples_split': [2, 5, 10]
}

grid = GridSearchCV(
    estimator=DecisionTreeClassifier(random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    verbose=1
)

grid.fit(x_encoded, y_encoded)

# ================================
# 5. RESULT
# ================================
print("=" * 50)
print("Best Parameters:", grid.best_params_)
print("Best Score:", grid.best_score_)
print("=" * 50)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Best model nikalo
best_model = grid.best_estimator_

# Train/Test split
x_train, x_test, y_train, y_test = train_test_split(
    x_encoded, y_encoded, test_size=0.2, random_state=42
)

# Best model train karo
best_model.fit(x_train, y_train)

# Prediction
y_pred = best_model.predict(x_test)

# Accuracy
train_acc = best_model.score(x_train, y_train)
test_acc = best_model.score(x_test, y_test)

print("=" * 50)
print(f"Train Accuracy: {train_acc:.4f}")
print(f"Test Accuracy:  {test_acc:.4f}")
print(f"Difference:     {train_acc - test_acc:.4f}")
print("=" * 50)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
import pandas as pd

# Saare results dekho
results = pd.DataFrame(grid.cv_results_)
results = results[['param_max_depth', 'param_criterion', 'param_min_samples_split', 'mean_test_score', 'std_test_score']]
results = results.sort_values('mean_test_score', ascending=False)

print("Top 10 Best Combinations:")
print(results.head(10).to_string(index=False))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.preprocessing import OneHotEncoder, LabelEncoder
import pandas as pd

# Data load
df = pd.read_csv("mushrooms.csv")
x = df.drop("type", axis=1)
y = df["type"]

# OneHot Encoding
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
x_onehot = encoder.fit_transform(x)

le_y = LabelEncoder()
y_encoded = le_y.fit_transform(y)

# Stratified K-Fold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Tuned Random Forest
rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    min_samples_split=2,
    random_state=42,
    n_jobs=-1
)

scores = cross_val_score(rf, x_onehot, y_encoded, cv=skf)
print("=== Random Forest (Tuned + OneHot + Stratified) ===")
print("CV Scores:", scores)
print("Mean:", scores.mean())
print("Std:", scores.std())
```
